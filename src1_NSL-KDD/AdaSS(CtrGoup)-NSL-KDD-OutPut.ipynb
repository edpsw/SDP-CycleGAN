```python
import numpy as np 
from pandas_ml import ConfusionMatrix
```

```python
import pandas as pd
from sklearn.metrics import roc_auc_score
  
def Sensitivity(TP,FN):
    sen = TP/(TP + FN)
    return sen

def Specificity(TN,FP):
    spe = TN/(TN + FP)
    return spe

def GMean(sen,spe):
    return (sen * spe) ** 0.5

def FPR(FP,TN):
    return FP/(FP + TN)
  
def Precision(TP, TN, P):
    return (TP + TN) / P
    
def Balance(fpr, pre):
    return 1 - ((fpr**2 + (1-pre)**2)/2) ** 2
    
def AUC(y_true, y_pred):
    return roc_auc_score(y_true, y_pred)

def ACC(TP,TN,P):
    return (TP+TN)/P

def TPR(TP,FN):
    return TP/(TP+FN)

def F1(pre,tpr):
    return 2 * pre * tpr / (pre + tpr)
    
def MyMetrics(cm):
    tab = pd.crosstab(cm._y_true, cm._y_pred)
    TN = tab.reindex()[0][0]
    TP = tab.reindex()[1][1]
    FN = tab.reindex()[0][1]
    FP = tab.reindex()[1][0]
    P = cm.population
    sen = Sensitivity(TP,FN)
    spe = Specificity(TN,FP)
    fpr = FPR(FP,TN)
    pre = Precision(TP, TN, P)
    tpr = TPR(TP,FN)
    acc = ACC(TP,TN,P)
    f1 = F1(pre,tpr)
    
    gmean = GMean(sen,spe)
    balance = Balance(fpr, pre)
    auc = AUC(cm._y_true, cm._y_pred)
    print("FPR:",fpr)
    print("Precision:",pre)
    print("Accuracy:",acc)
    print("TPR:",tpr)
    print("F1_score:",f1)
    
    print("Sensitivity:",sen)
    print("Specificity:",spe)
    print("G-Mean:",gmean)
    print("Balance:", balance)
    print("AUC:", auc)
```

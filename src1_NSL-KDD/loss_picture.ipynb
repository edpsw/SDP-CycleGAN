```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import pylab as pl
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
D_loss =np.loadtxt("./output/D_loss.csv") 
G_loss = np.loadtxt("./output/G_loss.csv",float,delimiter = ",",usecols = (0,1,2))

x = np.arange(20)
y = D_loss[0:100000:5000]
x1 = np.arange(len(G_loss[0:100000:5000,0]))
y1 = G_loss[0:100000:5000,0]
y2 = G_loss[0:100000:5000,1]
y3 = G_loss[0:100000:5000,2]

fig = plt.figure(figsize = (10,6))   
ax1 = fig.add_subplot(1, 1, 1) 

pl.plot(x,y,'g-',label=u'D_loss')
#p_1 = pl.plot(x1, y1,'r-', label = u'G_loss_1')
p_2 = pl.plot(x1, y2,'b-', label = u'GA_loss')
p_3 = pl.plot(x1, y3,'y-', label = u'GB_loss')
pl.legend()
pl.xlabel(u'epoch')
pl.ylabel(u'loss')
plt.title('Compare loss for D_loss and G_loss')
```
